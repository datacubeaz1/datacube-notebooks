```python
import sys
import pandas as pd

if sys.platform == "emscripten":      # JupyterLite (saytda)
    BASE = ""
else:                                 # Google Colab
    BASE = "https://raw.githubusercontent.com/datacubeaz1/datacube-notebooks/main/content/"

df = pd.read_csv(BASE + "datasets/users_behavior.csv")
df.shape
```

```python
import duckdb
duckdb.sql("SELECT is_ultra, count(*) AS n, avg(calls) AS avg_calls FROM df GROUP BY is_ultra ORDER BY 1").df()
```

```python
%pip install seaborn
import seaborn as sns
sns.histplot(df["calls"])
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
X=df.drop(columns="is_ultra"); y=df["is_ultra"]
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=0.4,random_state=12345)
DecisionTreeClassifier(max_depth=3).fit(Xtr,ytr).score(Xte,yte)
```
